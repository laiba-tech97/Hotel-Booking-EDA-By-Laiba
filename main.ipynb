Missing values graph

```python
df.isnull().sum().plot(kind="bar", figsize=(12,5))
plt.title("Missing Values")
plt.show()
```

Fill missing values

```python
df["children"] = df["children"].fillna(0)

df["country"] = df["country"].fillna(df["country"].mode()[0])

df["agent"] = df["agent"].fillna(0)

df["company"] = df["company"].fillna(0)
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df["reservation_status_date"] = pd.to_datetime(df["reservation_status_date"])
```

Outliers using box plots

```python
plt.figure(figsize=(10,5))

sns.boxplot(x=df['adr'], color='skyblue')

plt.title("Outliers in Average Daily Rate (ADR)")
plt.xlabel("ADR")
plt.show()
```

Outlier Graph – Lead Time

```python
plt.figure(figsize=(10,5))

sns.boxplot(x=df['lead_time'], color='orange')

plt.title("Outliers in Lead Time")
plt.xlabel("Lead Time")
plt.show()
```

Outlier Graph – Total Guests

```python
df['total_guests'] = df['adults'] + df['children'] + df['babies']

plt.figure(figsize=(10,5))

sns.boxplot(x=df['total_guests'], color='lightgreen')

plt.title("Outliers in Total Guests")
plt.xlabel("Total Guests")
plt.show()
```

Detect Outliers Using IQR

```python
Q1 = df['adr'].quantile(0.25)
Q3 = df['adr'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['adr'] < lower) | (df['adr'] > upper)]

print("Total Outliers:", len(outliers))
print(outliers[['adr']].head())
```

Q1. How many bookings are canceled and not canceled?

```python
df["is_canceled"].value_counts().plot(kind="bar")

plt.title("Canceled vs Not Canceled")
plt.show()

print(df["is_canceled"].value_counts())
```

Output: Shows the total number of canceled and non-canceled bookings. 0 means not canceled and 1 means canceled

Which hotel has the most bookings?

```python
df["hotel"].value_counts().plot(kind="bar")

plt.title("Bookings by Hotel")
plt.show()

print(df["hotel"].value_counts())
```

Output : City Hotel has more booking

Q3. What is the distribution of lead time?

```python
df["lead_time"].plot(kind="hist", bins=30)

plt.title("Lead Time")
plt.show()

print("Average Lead Time:", df["lead_time"].mean())
```

Result: Customers booked approximately 104 days before arrival on average.

Q4. Which market segment has the most bookings?

```python
df["market_segment"].value_counts().plot(kind="bar")

plt.title("Market Segment Bookings")
plt.show()

print(df["market_segment"].value_counts())
```

Result: Online TA has the highest bookings.

Q5. What is the relationship between lead time and ADR?

```python
df.plot(
    kind="scatter",
    x="lead_time",
    y="adr"
)

plt.title("Lead Time vs ADR")
plt.show()

print(df["lead_time"].corr(df["adr"]))
```

Result: There is a very weak negative correlation between lead time and ADR.

Q6. Does the number of special requests relate to cancellation?

```python
df.groupby("total_of_special_requests")["is_canceled"].mean().plot(kind="bar")

plt.title("Special Requests vs Cancellation")
plt.ylabel("Cancellation Rate")
plt.show()
```

Result: Bookings with zero special requests have the highest cancellation rate.

Q7. Which hotel has the highest average ADR?

```python
df.groupby("hotel")["adr"].mean().plot(kind="bar")

plt.title("Average ADR by Hotel")
plt.show()

print(df.groupby("hotel")["adr"].mean())
```

Result: City Hotel has the highest average ADR.

Q8. Which customer type has the highest cancellation rate?

```python
df.groupby("customer_type")["is_canceled"].mean().plot(kind="bar")

plt.title("Cancellation Rate by Customer Type")
plt.show()

print(df.groupby("customer_type")["is_canceled"].mean())
```

Output: Transient customers have the highest cancellation rate.
